```python
feuilles = [(1, "A"), (3, "A"), (4, "A"), (8, "B"), (10, "B"), (11, "B")]  # (longueur en cm, espèce)
nouvelle = 5
K = 3

# distance entre deux feuilles
def distance_euclidienne (a,b):
  return abs(b-a)

# algorithme Knn
def knn (donnees,x,k):
  # distance de x de feuille connue
  distances = []
  for longueur , espece in donnees:
    d=distance_euclidienne(x,longueur)
    distances.append((d,espece))
  # trier de la plus proche à la plus lointaine
  distances.sort()
  # garder les k premières
  voisins=distances[:k]
  # voter les voisins majoritairement
  votes = {}
  for d, espece in voisins:
    if espece in votes:
      votes[espece]+=1
    else:
      votes[espece]=1
  # l'espece qui a plus de votes
  gagnant = max(votes, key=votes.get)
  print("distances triées:" , distances)
  print("les",k,"voisins:",voisins)
  print("votes:",votes)
  return gagnant

def evaluer(donnees, k):
  bonnes = 0
  # matrice de confusion : (vraie espèce, espèce prédite) -> nombre
  confusion = {("A", "A"): 0, ("A", "B"): 0, ("B", "A"): 0, ("B", "B"): 0}

  for i in range(len(donnees)):
      # on cache la feuille i : elle devient la feuille « de test »
      longueur, vraie = donnees[i]
      # les autres feuilles servent d'entraînement
      entrainement = donnees[:i] + donnees[i + 1:]

      predite = knn(entrainement, longueur, k)
      confusion[(vraie, predite)] += 1
      if predite == vraie:
          bonnes += 1
          verdict = "juste"
      else:
          verdict = "FAUX"
      print(f"  feuille {longueur:>2} cm : vraie = {vraie}, prédite = {predite}  → {verdict}")

  accuracy = bonnes / len(donnees)
  print(f"  Accuracy = {bonnes}/{len(donnees)} = {accuracy:.2f}")
  print("  Matrice de confusion (lignes = vraie, colonnes = prédite) :")
  print("          A   B")
  print(f"     A   {confusion[('A', 'A')]}   {confusion[('A', 'B')]}")
  print(f"     B   {confusion[('B', 'A')]}   {confusion[('B', 'B')]}")
  return accuracy

for k in [1, 3, 5]:
    print(f"===== K = {k} =====")
    evaluer(feuilles, k)
    print()
```
